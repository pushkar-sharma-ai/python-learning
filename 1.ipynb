```python
x=df[["Year","Present_Price","Kms_Driven"]]
y=df["Selling_Price"]

print("X shape:", x.shape)
print("Y shape:" , y.shape)
```

```python
X_train,X_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
print("train rows:", len(X_train))
print("test rows:", len(X_test))
```

```python
model=LinearRegression()
model.fit(X_train, y_train)
print("correct")
```

```python
y_pred=model.predict(X_test)
compare=pd.DataFrame({"actually" : y_test.values,"predicted" : y_pred.round(2)})
compare["Error"] = (compare["actually"]-compare ["predicted"]).round(2)
compare.head(5)
```

```python
new_car=pd.DataFrame([{
    "Year" : 2015,
    "Present_Price" : 9.0,
    "Kms_Driven" : 30000

}])
price=model.predict(new_car)[0]
print(f"predict car price : {price:.2f} lakh")
```
